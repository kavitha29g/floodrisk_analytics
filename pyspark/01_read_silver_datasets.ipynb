```python

# Chennai Urban Intelligence Project
# Read Silver datasets using PySpark

silver_path = "/Volumes/workspace/default/chennai_silver_input/"

print("Silver path:", silver_path)
```

```python

# Read all 8 cleaned datasets

rainfall_df = spark.read.option("header", True).option("inferSchema", True).csv(silver_path + "rainfall_cleaned.csv")

flood_df = spark.read.option("header", True).option("inferSchema", True).csv(silver_path + "flood_cleaned.csv")

road_df = spark.read.option("header", True).option("inferSchema", True).csv(silver_path + "road_complaints_cleaned.csv")

traffic_df = spark.read.option("header", True).option("inferSchema", True).csv(silver_path + "traffic_cleaned.csv")

grievances_df = spark.read.option("header", True).option("inferSchema", True).csv(silver_path + "public_grievances_cleaned.csv")

waterlogging_df = spark.read.option("header", True).option("inferSchema", True).csv(silver_path + "waterlogging_locations_cleaned.csv")

population_df = spark.read.option("header", True).option("inferSchema", True).csv(silver_path + "population_cleaned.csv")

emergency_df = spark.read.option("header", True).option("inferSchema", True).csv(silver_path + "emergency_response_cleaned.csv")

print("All 8 datasets loaded successfully!")
```

```python

# Check row counts of all datasets

datasets = {
    "Rainfall": rainfall_df,
    "Flood": flood_df,
    "Road Complaints": road_df,
    "Traffic": traffic_df,
    "Public Grievances": grievances_df,
    "Waterlogging": waterlogging_df,
    "Population": population_df,
    "Emergency Response": emergency_df
}

for name, df in datasets.items():
    print(name, ":", df.count(), "rows")
```
