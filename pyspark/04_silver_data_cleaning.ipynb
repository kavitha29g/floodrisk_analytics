```python

rainfall_df = spark.table("workspace.default.bronze_rainfall")
flood_df = spark.table("workspace.default.bronze_flood")
road_df = spark.table("workspace.default.bronze_road_complaints")
waterlogging_df = spark.table("workspace.default.bronze_waterlogging")
traffic_df = spark.table("workspace.default.bronze_traffic")
emergency_df = spark.table("workspace.default.bronze_emergency_response")
grievances_df = spark.table("workspace.default.bronze_public_grievances")
population_df = spark.table("workspace.default.bronze_population")

print("All 8 Bronze datasets loaded successfully!")
```

```python

from pyspark.sql.functions import col, sum, when

rainfall_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in rainfall_df.columns
]).show()
```

```python

flood_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in flood_df.columns
]).show()
```
