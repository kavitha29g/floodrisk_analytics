```python

rainfall_df = spark.table("workspace.default.bronze_rainfall")
flood_df = spark.table("workspace.default.bronze_flood")
road_df = spark.table("workspace.default.bronze_road_complaints")
waterlogging_df = spark.table("workspace.default.bronze_waterlogging")
traffic_df = spark.table("workspace.default.bronze_traffic")
emergency_df = spark.table("workspace.default.bronze_emergency_response")
grievances_df = spark.table("workspace.default.bronze_public_grievances")
population_df = spark.table("workspace.default.bronze_population")

print("All 8 Bronze datasets loaded successfully!")
```

```python

from pyspark.sql.functions import col, sum, when

rainfall_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in rainfall_df.columns
]).show()
```

```python

flood_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in flood_df.columns
]).show()
```

```python

road_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in road_df.columns
]).show()


waterlogging_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in waterlogging_df.columns
]).show()



waterlogging_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in waterlogging_df.columns
]).show()


traffic_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in traffic_df.columns
]).show()


emergency_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in emergency_df.columns
]).show()


grievances_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in grievances_df.columns
]).show()



population_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in population_df.columns
]).show()
```

```python

# STEP 2: DUPLICATE RECORD CHECK FOR ALL 8 DATASETS

# 1. Rainfall
print("========== RAINFALL ==========")
print("Total records:", rainfall_df.count())
print("Duplicate records:", rainfall_df.count() - rainfall_df.dropDuplicates().count())


# 2. Flood
print("\n========== FLOOD ==========")
print("Total records:", flood_df.count())
print("Duplicate records:", flood_df.count() - flood_df.dropDuplicates().count())


# 3. Road Complaints
print("\n========== ROAD COMPLAINTS ==========")
print("Total records:", road_df.count())
print("Duplicate records:", road_df.count() - road_df.dropDuplicates().count())


# 4. Waterlogging
print("\n========== WATERLOGGING ==========")
print("Total records:", waterlogging_df.count())
print("Duplicate records:", waterlogging_df.count() - waterlogging_df.dropDuplicates().count())


# 5. Traffic
print("\n========== TRAFFIC ==========")
print("Total records:", traffic_df.count())
print("Duplicate records:", traffic_df.count() - traffic_df.dropDuplicates().count())


# 6. Emergency Response
print("\n========== EMERGENCY RESPONSE ==========")
print("Total records:", emergency_df.count())
print("Duplicate records:", emergency_df.count() - emergency_df.dropDuplicates().count())


# 7. Public Grievances
print("\n========== PUBLIC GRIEVANCES ==========")
print("Total records:", grievances_df.count())
print("Duplicate records:", grievances_df.count() - grievances_df.dropDuplicates().count())


# 8. Population
print("\n========== POPULATION ==========")
print("Total records:", population_df.count())
print("Duplicate records:", population_df.count() - population_df.dropDuplicates().count())
```
